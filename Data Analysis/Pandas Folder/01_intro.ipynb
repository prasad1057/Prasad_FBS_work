```python
import pandas as pd
```

```python
s1 = pd.Series([10, 20, 30, 40])
s1
```

```python
s2 = pd.Series({'a':1, 'b':2, 'c':3, 'd':4})
s2
```

```python
s3 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])            # abcd are the index of series
s3
```

```python
s3['a']
```

```python
s4 = pd.Series({'c':50, 'd':60, 'e':70, 'f':80})
s4
```

```python
s3 + s4
```

```python
s3.max()
```

```python
s3.min()
```

```python
s4.mode()           #mode : high occurances values
```

```python
s4.var()            # var : square of standard deviation
```
