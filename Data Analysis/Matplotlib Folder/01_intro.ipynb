#### LinePlot
1. Showing the trends.
2. Time series analysis.

```python
overs = np.arange(1, 21)
team1 = np.random.randint(0, 37, 20)
team2 = np.random.randint(0, 37, 20)
team1, team2
```

```python
team1_score = np.cumsum(team1)      #cumsum means add runs over by over and give total run at end
team2_score = np.cumsum(team2)

team1_score,team2_score
```

```python
plt.figure(figsize=(6,4))
plt.plot(overs, team1_score, color='r', marker='o', label='Team 1')
plt.plot(overs, team2_score, color='b', marker='*', label='Team 2')
plt.grid(alpha=0.3, linestyle='--')
plt.xticks(overs)               #to want integer value in x axis
plt.legend()
plt.title('IPL Match Score Comaprison')
plt.xlabel('Overs')
plt.ylabel('Runs')

plt.savefig('IPL match score comparison.jpg')
plt.show()
```

#### ScatterPlot
1. Relationship between to variables
2. Corelation analysis

```python
plt.figure(figsize=(6,4))
plt.scatter(overs, team1, label='Team 1', color='r', marker='>', edgecolors='g', s=100)
plt.scatter(overs, team2, label='Team 2', color='b', marker='*', edgecolors='orange', s=100)
plt.grid(alpha=0.3, linestyle='--')
plt.xticks(overs)
plt.legend()
plt.show()
```

#### Pie Chart

2. Part-to-whole comparision

```python
classes = ['Python', 'Java', 'ML', 'DS', "Testing"]
class1 = [45, 15, 35, 24, 30]
colors = ['c', 'b', 'r', 'y', 'g']
explode = [0.03, 0, 0.1, 0, 0]
plt.pie(class1, labels = classes, colors= colors, explode=explode)
plt.show()
```
